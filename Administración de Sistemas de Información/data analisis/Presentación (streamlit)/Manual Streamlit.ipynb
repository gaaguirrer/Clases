## 1.1 Instalación y Configuración

### Instalación de Streamlit

Streamlit es una biblioteca de Python que permite crear aplicaciones web interactivas de manera sencilla. Para instalar Streamlit, abre tu terminal y ejecuta:

```bash
pip install streamlit
```

Para verificar que la instalación fue exitosa, ejecuta:

```bash
streamlit --version
```

### Estructura del Proyecto

Crea una carpeta para tu proyecto y dentro de ella crea un archivo llamado `app.py`. Esta será tu aplicación principal.

```
galeria-arte/
 app.py
 .streamlit/
     config.toml  (lo crearemos más adelante)
```

### ¿Cómo ejecutar una aplicación Streamlit?

A diferencia de los scripts Python normales que se ejecutan con `python app.py`, las aplicaciones Streamlit se ejecutan con:

```bash
streamlit run app.py
```

Este comando iniciará un servidor web local y abrirá automáticamente tu navegador con la aplicación.

### Elementos de Texto Básicos

Streamlit ofrece varios métodos para mostrar texto con diferentes jerarquías y formatos:

````python
import streamlit as st

# st.title() crea un título principal (H1 en HTML)
# Es el texto más grande y debe usarse solo una vez por página
st.title('Galería de Arte Digital')

# st.header() crea un encabezado de sección (H2 en HTML)
# Úsalo para dividir tu aplicación en secciones principales
st.header('Bienvenido a Nuestra Colección')

# st.subheader() crea un subtítulo (H3 en HTML)
# Úsalo para subsecciones dentro de cada header
st.subheader('Arte Contemporáneo')

# st.text() muestra texto simple sin formato
# Es útil para texto plano que no necesita estilo
st.text('Explora nuestra colección de obras de arte modernas.')

# st.write() puede mostrar múltiples tipos de datos
# Es el método más flexible de Streamlit
st.write('Puedes usar **negritas** y *cursivas* con st.write()')
````

### Formato con Markdown

Streamlit soporta Markdown, lo que te permite crear texto con formato rico:

````python
import streamlit as st

# st.markdown() permite usar la sintaxis completa de Markdown
# Markdown es un lenguaje de marcado ligero para dar formato a texto
st.markdown("""
### Características de nuestra galería:

- **Colección extensa**: Más de 100 obras de arte
- **Artistas diversos**: De todo el mundo
- **Alta calidad**: Imágenes en resolución 4K

---

*Disfruta explorando el arte digital*

> "El arte es la mentira que nos permite comprender la verdad" - Pablo Picasso
""")

# También puedes usar st.markdown() para texto en línea
st.markdown('Este texto tiene **negritas**, *cursivas* y `código`')
````

### Mostrar Imágenes

`st.image()` permite mostrar imágenes desde URLs o archivos locales:

````python
import streamlit as st

# st.image() muestra imágenes en tu aplicación
# Puedes usar URLs de internet o rutas de archivos locales
st.image(
    'https://images.unsplash.com/photo-1579783902614-a3fb3927b6a5?w=800',  # URL de la imagen
    caption='Arte Abstracto Digital',  # Texto que aparece debajo de la imagen
    width=600  # Ancho de la imagen en píxeles (opcional)
)

# Ejemplo con imagen local (si tienes una imagen en tu carpeta)
# st.image('ruta/a/tu/imagen.jpg', caption='Mi obra de arte', width=500)
````

### Separadores Visuales

`st.divider()` crea una línea horizontal para separar secciones:

````python
import streamlit as st

st.header('Sección 1')
st.write('Contenido de la primera sección')

# st.divider() crea una línea horizontal de separación
# Es útil para dividir visualmente diferentes secciones de contenido
st.divider()

st.header('Sección 2')
st.write('Contenido de la segunda sección')
````

### Mostrar Código

`st.code()` muestra bloques de código con resaltado de sintaxis:

````python
import streamlit as st

st.subheader('Ejemplo de Código Python')

# st.code() muestra código con resaltado de sintaxis
# El parámetro 'language' especifica el lenguaje de programación
st.code("""
def saludar(nombre):
    return f"Hola, {nombre}!"

mensaje = saludar("Artista")
print(mensaje)
""", language='python')  # Especificamos que es código Python

# También funciona con otros lenguajes
st.code('const mensaje = "Hola desde JavaScript";', language='javascript')
````

## 2.1 Sidebar - Barra Lateral

El sidebar es un panel lateral que permanece visible y es perfecto para navegación o controles.

### ¿Qué es el Sidebar?

El sidebar de Streamlit es una barra lateral que se muestra a la izquierda de la aplicación. Es ideal para:
- Menús de navegación
- Filtros y controles
- Información contextual
- Opciones de configuración


````python
import streamlit as st

# Todos los elementos que agregues con st.sidebar aparecerán en la barra lateral
# Título del sidebar
st.sidebar.title('Navegación')

# Texto en el sidebar
st.sidebar.write('Usa este menú para navegar')

# Separador en el sidebar
st.sidebar.divider()

# Sección en el sidebar
st.sidebar.header('Secciones')

# Botones en el sidebar
# Cada botón puede triggear diferentes acciones
if st.sidebar.button('Inicio'):
    st.write('Has hecho clic en Inicio')

if st.sidebar.button('Galería'):
    st.write('Has hecho clic en Galería')

if st.sidebar.button('Contacto'):
    st.write('Has hecho clic en Contacto')

# Contenido principal fuera del sidebar
st.title('Página Principal')
st.write('El contenido principal aparece aquí')
````

### Navegación con Session State

Para crear una navegación real entre secciones, usaremos `st.session_state`:

````python
import streamlit as st

# Inicializar el estado de la página si no existe
# st.session_state nos permite mantener variables entre rerenderizados
if 'pagina_actual' not in st.session_state:
    st.session_state.pagina_actual = 'Inicio'

# Sidebar con navegación
st.sidebar.title('Menú de Navegación')

# Crear botones que cambien el valor en session_state
if st.sidebar.button('Inicio'):
    st.session_state.pagina_actual = 'Inicio'

if st.sidebar.button('Galería'):
    st.session_state.pagina_actual = 'Galería'

if st.sidebar.button('Sobre Nosotros'):
    st.session_state.pagina_actual = 'Sobre Nosotros'

# Mostrar contenido según la página seleccionada
if st.session_state.pagina_actual == 'Inicio':
    st.title('Bienvenido a la Galería')
    st.write('Esta es la página de inicio')
    
elif st.session_state.pagina_actual == 'Galería':
    st.title('Nuestra Colección')
    st.write('Aquí se mostrarán las obras de arte')
    
elif st.session_state.pagina_actual == 'Sobre Nosotros':
    st.title('Sobre Nosotros')
    st.write('Información sobre nuestra galería')
````

## 2.2 Containers - Contenedores

`st.container()` permite agrupar elementos y controlar su orden de renderizado.


````python
import streamlit as st

# st.container() crea un contenedor donde puedes agrupar elementos
# Los contenedores ayudan a organizar y controlar el flujo de tu aplicación

# Crear un contenedor para el encabezado
header_container = st.container()

# Crear un contenedor para el contenido
content_container = st.container()

# Crear un contenedor para el pie de página
footer_container = st.container()

# Ahora podemos añadir contenido a cada contenedor
with header_container:
    st.title('Mi Aplicación')
    st.write('Subtítulo de la aplicación')

with content_container:
    st.header('Contenido Principal')
    st.write('Aquí va el contenido principal de la aplicación')
    st.write('Puedes añadir múltiples elementos dentro del contenedor')

with footer_container:
    st.divider()
    st.text('Copyright 2024')
````

### st.empty() - Placeholders Dinámicos

`st.empty()` crea un placeholder que puede ser sobrescrito dinámicamente:

````python
import streamlit as st
import time

# st.empty() crea un contenedor vacío que puede ser reemplazado
# Esto es útil para contenido que cambia dinámicamente

# Crear un placeholder vacío
placeholder = st.empty()

# Botón para iniciar la simulación
if st.button('Iniciar contador'):
    # Este contenido irá reemplazando lo que está en el placeholder
    for i in range(5):
        # Cada iteración reemplaza completamente el contenido anterior
        placeholder.write(f'Contador: {i + 1}')
        time.sleep(1)  # Esperar 1 segundo
    
    # Al final, mostrar mensaje de completado
    placeholder.success('Contador completado!')
````

## 2.3 Layouts de Columnas

`st.columns()` permite crear diseños de múltiples columnas.


````python
import streamlit as st

# st.columns() crea columnas para organizar contenido horizontalmente
# Puedes especificar el número de columnas o proporciones personalizadas

st.header('Layout de 2 Columnas')

# Crear 2 columnas de igual tamaño
col1, col2 = st.columns(2)

# Añadir contenido a la columna 1 using 'with'
with col1:
    st.subheader('Columna 1')
    st.write('Contenido de la primera columna')
    st.image('https://images.unsplash.com/photo-1579783902614-a3fb3927b6a5?w=400',
             use_column_width=True)

# Añadir contenido a la columna 2
with col2:
    st.subheader('Columna 2')
    st.write('Contenido de la segunda columna')
    st.image('https://images.unsplash.com/photo-1541961017774-22349e4a1262?w=400',
             use_column_width=True)
````

### Columnas con Proporciones Personalizadas


````python
import streamlit as st

st.header('Columnas con Proporciones Diferentes')

# Crear 3 columnas con proporciones 2:3:1
# La columna 1 ocupará 2 partes, columna 2 ocupará 3 partes, columna 3 ocupará 1 parte
col1, col2, col3 = st.columns([2, 3, 1])

with col1:
    st.write('Columna pequeña (2 partes)')
    
with col2:
    st.write('Columna grande (3 partes)')
    
with col3:
    st.write('Columna muy pequeña (1 parte)')
````

## 2.4 Tabs - Pestañas

`st.tabs()` organiza contenido en pestañas para mejor navegación.


````python
import streamlit as st

# st.tabs() crea pestañas para organizar contenido
# El usuario puede hacer clic en cada pestaña para ver su contenido

st.header('Navegación con Pestañas')

# Crear 3 pestañas con nombres descriptivos
tab1, tab2, tab3 = st.tabs(['Arte Moderno', 'Arte Clásico', 'Arte Abstracto'])

# Contenido de la primera pestaña
with tab1:
    st.subheader('Colección de Arte Moderno')
    st.write('Obras de arte del siglo XX y XXI')
    st.image('https://images.unsplash.com/photo-1579783902614-a3fb3927b6a5?w=600',
             caption='Ejemplo de arte moderno')

# Contenido de la segunda pestaña
with tab2:
    st.subheader('Colección de Arte Clásico')
    st.write('Obras maestras de los grandes maestros')
    st.write('Aquí iría contenido sobre arte clásico')

# Contenido de la tercera pestaña
with tab3:
    st.subheader('Colección de Arte Abstracto')
    st.write('Expresiones no figurativas del arte')
    st.write('Contenido sobre arte abstracto')
````

## 2.5 Expander - Contenido Colapsable

`st.expander()` crea secciones que se pueden expandir/colapsar.


````python
import streamlit as st

# st.expander() crea una sección colapsable
# Es útil para ocultar información adicional que no todos necesitan ver

st.header('Información con Expanders')

# Crear un expander con título
with st.expander('Ver detalles de la obra'):
    st.write('Esta es información adicional sobre la obra de arte')
    st.write('Artista: Desconocido')
    st.write('Año: 2023')
    st.write('Técnica: Digital')

# Crear otro expander
with st.expander('Información técnica'):
    st.write('Resolución: 4K')
    st.write('Formato: PNG')
    st.code('dimensiones = (3840, 2160)', language='python')

# Expander expandido por defecto
with st.expander('Galería de imágenes', expanded=True):
    # expanded=True hace que el expander esté abierto por defecto
    st.write('Aquí se mostrarían las imágenes')
    st.image('https://images.unsplash.com/photo-1541961017774-22349e4a1262?w=600')
````

## 2.6 Proyecto: Versión 2 - Galería con Navegación

Ahora crearemos la Versión 2 de nuestra Galería de Arte Digital incorporando todo lo aprendido en este módulo.

**Características de esta versión:**
- Sidebar con navegación
- Múltiples secciones organizadas
- Layouts de columnas para mostrar obras
- Tabs para categorizar el arte
- Expanders para información adicional

### `st.text_input()` - Entrada de Texto Simple


````python
import streamlit as st

# st.text_input() crea un campo de entrada de texto de una sola línea
# El valor ingresado se almacena en la variable

# Campo de texto básico
nombre = st.text_input('Ingresa tu nombre')

# Mostrar el valor ingresado
if nombre:
    st.write(f'Hola, {nombre}!')

# Campo de texto con valor por defecto
ciudad = st.text_input('Tu ciudad', value='Madrid')
st.write(f'Ciudad seleccionada: {ciudad}')

# Campo de texto con placeholder (texto de ayuda)
email = st.text_input('Email', placeholder='correo@ejemplo.com')

# Campo de texto tipo password (oculta el texto)
password = st.text_input('Contraseña', type='password')
# El texto ingresado se muestra como asteriscos
````

## 3.2 Inputs Numéricos y de Fecha

### `st.number_input()` - Entrada Numérica


````python
import streamlit as st

# st.number_input() crea un campo para ingresar números
# Incluye botones +/- para incrementar/decrementar

# Número entero básico
edad = st.number_input('Edad', min_value=0, max_value=120, value=25)
st.write(f'Tienes {edad} años')

# Número decimal (float)
precio = st.number_input(
    'Precio de la obra',
    min_value=0.0,
    max_value=100000.0,
    value=1500.50,
    step=0.10,  # Incremento de 0.10 cada vez
    format='%.2f'  # Formato con 2 decimales
)
st.write(f'Precio: ${precio:.2f}')

# Número sin límites
cantidad = st.number_input('Cantidad', value=1, step=1)
````

### `st.slider()` - Control Deslizante


````python
import streamlit as st

# st.slider() crea un control deslizante para seleccionar valores
# Es más visual e intuitivo que number_input

# Slider básico
puntuacion = st.slider('Califica esta obra', 0, 10, 5)
# min=0, max=10, valor_inicial=5
st.write(f'Puntuación: {puntuacion}/10')

# Slider con decimales
brillo = st.slider('Ajustar brillo', 0.0, 2.0, 1.0, 0.1)
# paso de 0.1
st.write(f'Brillo: {brillo}x')

# Slider de rango (seleccionar min y max)
rango_precio = st.slider(
    'Rango de precio',
    0, 10000,
    (1000, 5000)  # Valor inicial es una tupla (min, max)
)
st.write(f'Rango seleccionado: ${rango_precio[0]} - ${rango_precio[1]}')

# Slider con etiquetas personalizadas
import datetime
fecha_slider = st.slider(
    'Selecciona fecha',
    min_value=datetime.date(2020, 1, 1),
    max_value=datetime.date(2024, 12, 31),
    value=datetime.date(2023, 6, 1)
)
st.write(f'Fecha: {fecha_slider}')
````

### `st.date_input()` y `st.time_input()`


````python
import streamlit as st
import datetime

# st.date_input() muestra un calendario para seleccionar fechas
fecha = st.date_input(
    'Fecha de creación de la obra',
    value=datetime.date.today()  # Valor por defecto: hoy
)
st.write(f'Fecha seleccionada: {fecha}')

# Seleccionar rango de fechas
rango_fechas = st.date_input(
    'Período de exposición',
    value=(
        datetime.date(2024, 1, 1),
        datetime.date(2024, 12, 31)
    )
)
if len(rango_fechas) == 2:
    st.write(f'Desde: {rango_fechas[0]} hasta: {rango_fechas[1]}')

# st.time_input() para seleccionar hora
hora = st.time_input(
    'Hora de apertura',
    value=datetime.time(9, 0)  # 09:00 AM
)
st.write(f'Hora: {hora}')
````

### `st.multiselect()` - Selección Múltiple


````python
import streamlit as st

# st.multiselect() permite seleccionar MÚLTIPLES opciones
# Devuelve una lista con todas las opciones seleccionadas

# Multiselect básico
colores = st.multiselect(
    'Selecciona colores predominantes',
    ['Rojo', 'Azul', 'Verde', 'Amarillo', 'Negro', 'Blanco']
)
st.write(f'Colores seleccionados: {colores}')
st.write(f'Total: {len(colores)} colores')

# Multiselect con valores por defecto
tecnicas = st.multiselect(
    'Técnicas utilizadas',
    ['Óleo', 'Acrílico', 'Acuarela', 'Digital', 'Mixta'],
    default=['Digital', 'Mixta']  # Valores preseleccionados
)

# Usar las selecciones para filtrar
if tecnicas:
    st.info(f'Mostrando obras con: {', '.join(tecnicas)}')
else:
    st.warning('No has seleccionado ninguna técnica')
````

### `st.button()` - Botones


````python
import streamlit as st

# st.button() crea un botón clickeable
# Devuelve True cuando se hace clic, False en caso contrario

# Botón básico
if st.button('Guardar'):
    st.success('Obra guardada correctamente!')

# Botón con ancho completo
if st.button('Compartir', use_container_width=True):
    st.info('Compartiendo en redes sociales...')

# Botones en columnas
col1, col2, col3 = st.columns(3)

with col1:
    if st.button('Me gusta', use_container_width=True):
        st.write('Te gusta esta obra!')

with col2:
    if st.button('Comentar', use_container_width=True):
        st.write('Deja tu comentario')

with col3:
    if st.button('Reportar', use_container_width=True):
        st.write('Obra reportada')

# IMPORTANTE: Los botones no mantienen su estado
# Solo son True en el momento del clic
````

### `st.checkbox()` - Casillas de Verificación


````python
import streamlit as st

# st.checkbox() crea una casilla de verificación
# Devuelve True si está marcada, False si no
# A diferencia de button(), mantiene su estado

# Checkbox básico
mostrar_detalles = st.checkbox('Mostrar detalles de la obra')

if mostrar_detalles:
    st.write('**Título:** Arte Digital Moderno')
    st.write('**Artista:** Juan Pérez')
    st.write('**Año:** 2024')
    st.write('**Precio:** $1,500')

# Checkbox con valor inicial True
acepto_terminos = st.checkbox('Acepto los términos y condiciones', value=True)

# Múltiples checkboxes para filtros
st.subheader('Filtros')
filtro_disponible = st.checkbox('Solo obras disponibles')
filtro_precio = st.checkbox('Filtrar por precio')
filtro_artista = st.checkbox('Buscar por artista')

# Aplicar filtros
filtros_activos = []
if filtro_disponible:
    filtros_activos.append('Disponibles')
if filtro_precio:
    filtros_activos.append('Precio')
if filtro_artista:
    filtros_activos.append('Artista')

if filtros_activos:
    st.write(f'Filtros aplicados: {', '.join(filtros_activos)}')
````

## 3.6 Session State con Widgets

### Uso del parámetro `key`

Todos los widgets interactivos de Streamlit aceptan un parámetro `key`. Al asignar una clave a un widget, su valor se guarda automáticamente en `st.session_state`.


````python
import streamlit as st

st.header('Session State y Widgets')

# Al dar una 'key' al widget, Streamlit automáticamente guarda su valor
st.text_input('Ingresa tu nombre', key='nombre_usuario')

# Podemos acceder al valor en cualquier parte de la aplicación
if 'nombre_usuario' in st.session_state and st.session_state.nombre_usuario:
    st.write(f"Valor en session_state: {st.session_state.nombre_usuario}")

# También podemos inicializar valores antes de que el widget se renderice
if 'contador' not in st.session_state:
    st.session_state.contador = 0

def incrementar():
    st.session_state.contador += 1

# Usar callbacks (on_click) para ejecutar funciones al interactuar
st.button('Incrementar Contador', on_click=incrementar)

st.write(f"Valor del contador: {st.session_state.contador}")
````

````python
# ========================================
# GALERÍA DE ARTE DIGITAL - VERSIÓN 3
# Formulario Interactivo
# ========================================

import streamlit as st

st.set_page_config(page_title='Agregar Obra', layout='centered')

st.title('Registro de Nueva Obra')
st.write('Completa el formulario para registrar tu arte en la galería.')

# Container para el formulario
form_container = st.container()

with form_container:
    col1, col2 = st.columns(2)
    
    with col1:
        titulo = st.text_input('Título de la obra', placeholder='Ej. La Noche Estrellada')
        estilo = st.selectbox('Estilo', ['Abstracto', 'Moderno', 'Clásico', 'Digital', 'Surrealista'])
        precio = st.number_input('Precio Estimado ($)', min_value=0.0, step=50.0)
    
    with col2:
        artista = st.text_input('Nombre del artista', placeholder='Tu nombre o seudónimo')
        fecha = st.date_input('Fecha de creación')
        color_base = st.color_picker('Color predominante', '#1E88E5')

    descripcion = st.text_area('Descripción detallada', height=100)
    
    # Términos y condiciones
    st.divider()
    acepta_terminos = st.checkbox('Confirmo que esta obra es original y tengo los derechos sobre ella.')
    
    # Botón de envío
    enviado = st.button('Registrar Obra', use_container_width=True)
    
    if enviado:
        if not acepta_terminos:
            st.error('Debes aceptar los términos y condiciones para continuar.')
        elif len(titulo) == 0 or len(artista) == 0:
            st.warning('El título y el nombre del artista son obligatorios.')
        else:
            st.success('Exito: La obra ha sido registrada correctamente.')
            
            # Mostrar resumen
            st.info('Resumen del registro:')
            st.write(f"**Título:** {titulo}")
            st.write(f"**Artista:** {artista}")
            st.write(f"**Estilo:** {estilo}")
            st.write(f"**Precio:** ${precio}")
````

## 4.1 DataFrames: Tablas Estáticas vs Tablas Interactivas

A menudo los estudiantes usan la función `st.write()` para mostrar datos tabulares sin saber que existen opciones dedicadas que otorgan mejor rendimiento y UX.

### Diferencias Clave:

1. **`st.dataframe(df)`**: Es la opción interactiva por defecto.
   - **Cuándo usar:** Cuando el usuario necesita explorar grandes datasets, ordenar columnas, agrandar la tabla o buscar texto dentro de los datos.
   - **Buenas prácticas:** Si el dataframe tiene 10,000+ filas, no lo envíes crudo al navegador todo el tiempo, usa filtros previos en un sidebar o limita los resultados (ej. `df.head(100)`).
   
2. **`st.table(df)`**: Es estática, dibuja una tabla HTML entera en el navegador.
   - **Cuándo usar:** Cuando el DataFrame es muy pequeño (ej. 10 filas resumen).
   - **MALAS PRÁCTICAS:** Jamás uses `st.table` para datasets masivos; congelarás la pestaña del navegador del usuario.

3. **`st.data_editor(df)`**: Permite edición in-situ.
   - **Cuándo usar:** Cuando necesitas que los usuarios corrijan registros o añadan información que luego enviarás de vuelta a la base de datos.


````python
import streamlit as st
import pandas as pd
import numpy as np

st.header('Demostración: Tablas')

# Construcción de un Dataset robusto
fechas = pd.date_range(start='2024-01-01', periods=10)
datos = pd.DataFrame({
    'Fecha': fechas,
    'Ganancias': np.random.randint(1000, 5000, size=10),
    'Gastos': np.random.randint(500, 2000, size=10),
    'Estado': ['Activo', 'Pausado', 'Activo', 'Cerrado', 'Pausado', 'Activo', 'Activo', 'Cerrado', 'Activo', 'Pausado']
})

# Agregando interactividad en st.dataframe con la api ColumnConfig
st.subheader("1. st.dataframe (Recomendado para exploración)")
st.write("La tabla debajo permite ordenación mediante clic en las cabeceras.")
st.dataframe(
    datos,
    use_container_width=True,
    hide_index=True,  # Ocultamos el index nativo para limpiar UI
    column_config={
        "Ganancias": st.column_config.NumberColumn(
            "Ingresos Brutos",
            format="$ %d",
        )
    }
)

st.subheader("2. st.data_editor (Tablas Editables)")
st.write("Doble-clic en la columna Estado para cambiar los valores.")
datos_editados = st.data_editor(
    datos,
    use_container_width=True,
    num_rows="dynamic" # Permite que el usuario añada filas
)

````

## 4.2 Métricas de Rendimiento (`st.metric`)

Las métricas son los "Key Performance Indicators" (KPIs) de tu aplicación.

- Usa el parámetro `delta_color` inteligentemente. Normalmente, una subida (verde) es buena ('normal'), pero si mides "Errores del Servidor", un aumento debería colorearse en rojo (`delta_color='inverse'`).


````python
import streamlit as st

st.header('Métricas con Comportamientos de Color')

# Dividimos usando la proporción de pantalla
kpi1, kpi2, kpi3 = st.columns(3)

with kpi1:
    st.metric(
        label="Nuevos Suscriptores",
        value="1,245",
        delta="154", 
        delta_color="normal" # ↑ Verde (Positivo)
    )

with kpi2:
    st.metric(
        label="Costos de Servidor",
        value="$540.00",
        delta="$45.00",
        delta_color="inverse" # ↑ Rojo (Aumento de costos es negativo)
    )

with kpi3:
    st.metric(
        label="Latencia Promedio",
        value="120ms",
        delta="-5ms",
        delta_color="inverse" # ↓ Verde (Menos latencia es mejor)
    )

````

## 5.3 Estructura y "Cards" Interactivas con CSS Puro

**La Regla de Oro del CSS en Streamlit:**
Nunca trates de forzar overrides a las clases internas (como `.st-bx` o `.css-1d391kg`), ya que estas clases son autogeneradas por React y cambiarán en cada versión de Streamlit, rompiendo tu código.

````python
import streamlit as st

st.header("Diseñando Componentes Complejos: Cards")

# Definiremos un bloque CSS en un string multilinea.
custom_css = """
<style>
/* Clase para la tarjeta principal */
.my-card {
    background-color: #ffffff;
    border-radius: 10px;
    padding: 20px;
    box-shadow: 0 4px 6px rgba(0, 0, 0, 0.1); /* Sombra elegante */
    margin-bottom: 20px;
    border-left: 5px solid #FF4B4B; /* Borde izquierdo distintivo */
    transition: transform 0.3s ease; /* Transición suave */
}

/* Efecto visual al pasar el ratón (Hover) */
.my-card:hover {
    transform: translateY(-5px); /* Se levanta un poco */
    box-shadow: 0 6px 12px rgba(0, 0, 0, 0.15);
}

.my-card-title {
    margin-top: 0;
    font-size: 1.25rem;
    font-weight: 600;
    color: #31333F;
}

.my-card-text {
    color: #555555;
    font-size: 1rem;
    line-height: 1.5;
}

/* Soporte para modo oscuro */
@media (prefers-color-scheme: dark) {
    .my-card {
        background-color: #262730;
        color: #FAFAFA;
        box-shadow: 0 4px 6px rgba(0, 0, 0, 0.5);
    }
    .my-card-title {
        color: #FAFAFA;
    }
    .my-card-text {
        color: #CCCCCC;
    }
}
</style>
"""

# Inyectamos los estilos
st.markdown(custom_css, unsafe_allow_html=True)

# Creamos las tarjetas usando la estructura CSS anterior
tarjeta_html = """
<div class="my-card">
    <h3 class="my-card-title">Resumen Financiero</h3>
    <p class="my-card-text">
        El balance del último trimestre muestra un incremento estable en 
        las operaciones generales, reduciendo costos pasivos en un 14%.
    </p>
</div>
"""

st.markdown(tarjeta_html, unsafe_allow_html=True)
````

## 5.4 CSS Externo: Manteniendo tu Código Limpio

A medida que tu aplicación crece y agregas más estilos (como hicimos en la sección anterior para las Cards), inyectar largos bloques de texto `<style>...</style>` directamente en tu archivo `app.py` hace que el código sea difícil de leer y mantener.

**La Mejor Práctica:** Crear un documento CSS separado (por ejemplo, `style.css`) y leerlo desde Python para inyectarlo en Streamlit.


````python
import streamlit as st
import os

st.header("Importando un archivo .css externo")

# 1. Supongamos que tienes un archivo llamado 'estilos.css' en la misma carpeta
# con el siguiente contenido:
# .alerta-personalizada { 
#     color: white; 
#     background-color: darkred; 
#     padding: 10px; 
#     border-radius: 5px; 
# }

# 2. Creamos una función en Python para leer el archivo .css
def aplicar_css_local(nombre_archivo):
    # Verificamos que el archivo exista para evitar errores
    if os.path.exists(nombre_archivo):
        with open(nombre_archivo, 'r') as f:
            css = f.read()
            # Inyectamos el contenido rodeado por etiquetas <style>
            st.markdown(f'<style>{css}</style>', unsafe_allow_html=True)
    else:
        st.warning(f"No se encontró el archivo: {nombre_archivo}")

# 3. Llamamos a la función al inicio de nuestra aplicación
aplicar_css_local('estilos.css')

# 4. Usamos las clases definidas en el archivo CSS externo
st.markdown('<div class="alerta-personalizada">¡Esta es una alerta estilizada desde un archivo externo!</div>', unsafe_allow_html=True)

````

## 6.1 Formularios Seguros (`st.form`)

**¿Por qué usarlos?** 
Streamlit re-ejecuta todo el script desde el principio cada vez que el usuario interactúa con un widget. En formularios largos, esto es ineficiente y molesto para el usuario. `st.form` agrupa los inputs y solo realiza el re-renderizado cuando se presiona el botón de envío (`st.form_submit_button`).

````python
import streamlit as st

st.header("Registro de Obras")

# Uso de st.form para evitar re-renderizados innecesarios por tecla pulsada
with st.form(key="formulario_registro"):
    titulo = st.text_input("Título de la obra")
    artista = st.text_input("Nombre del Artista")
    anio = st.number_input("Año de Creación", min_value=1900, max_value=2024, value=2023)
    es_original = st.checkbox("Certificado de Originalidad Incluido")
    
    # El botón de envío es obligatorio dentro de un formulario
    submit = st.form_submit_button("Registrar Obra en el Sistema")

if submit:
    if not titulo or not artista:
        st.error("Por favor completa los campos principales.")
    else:
        st.success(f"Obra '{titulo}' registrada correctamente en el sistema.")
````

## 6.2 El Peligro del Re-Renderizado y la Solución: `st.form()`

**Problema:** En Streamlit, interactuar con cualquier widget (ej. escribir una letra en `st.text_input`) provoca que TODO EL SCRIPT Python se vuelva a ejecutar de arriba hacia abajo de inmediato. 
Si tienes modelos de Machine Learning o lecturas de base de datos costosas debajo, esto congelará la aplicación por cada tecla presionada.

**Solución Institucional:** `st.form`. Todo lo que esté envuelto en un `form` "congela" su estado y no provoca la re-ejecución del script hasta que el usuario hace clic expresamente en un `st.form_submit_button`.

**Comportamientos a Evitar:**
- Nunca coloques código que te transporte a otra pestaña dentro de un submit de formulario, puede generar inconsistencias en session_state.
- Nunca metas otros `st.button` dentro de un form.


````python
import streamlit as st
import time

st.header("Formularios por Lotes (Batching)")

# Creamos el contenedor del form
with st.form(key='registro_avanzado', clear_on_submit=True):
    st.write("Ingresa los datos del nuevo producto:")
    cod_producto = st.text_input("Código SKU")
    descripcion = st.text_area("Descripción Comercial")
    
    col1, col2 = st.columns(2)
    with col1:
        categoria = st.selectbox("Categoría", ["Electrónica", "Materiales", "Servicios"])
    with col2:
        stock_minimo = st.number_input("Stock Mínimo Alerta", min_value=1)
        
    # Este es el único botón que puede vivir dentro de un st.form
    submit = st.form_submit_button("Sincronizar Producto")

if submit:
    if len(cod_producto) == 0:
        st.error("Es obligatorio adjuntar un SKU")
    else:
        st.success(f"Producto '{cod_producto}' sincronizado exitosamente en {categoria}.")
````

## 6.3 Manejo de Estado (`st.spinner`) y Caché (`@st.cache_data`)

Cuando lidiamos con procesamiento pesado (consultas a API o limpieza de datasets masivos), no queremos que el usuario espere sin feedback, ni queremos recalcular los mismos datos cada vez que clica un botón.


````python
import streamlit as st
import time
import pandas as pd

st.subheader("Simulación de Carga Pesada")

# Decorador de caché: guarda el resultado de esta función en la memoria.
# Si los parámetros de entrada no cambian, devuelve el resultado guardado
# inmediatamente en lugar de ejecutar la función.
@st.cache_data
def cargar_datos_ventas():
    # Simulamos una consulta lenta a una base de datos
    time.sleep(3)  # Suspensión intencional de 3 segundos
    
    data = {"Mes": ["Enero", "Febrero", "Marzo"], "Ventas_USD": [15000, 22000, 18500]}
    return pd.DataFrame(data)

# Usar st.spinner para proveer feedback visual (sin el uso de caracteres del sistema)
with st.spinner("Conectando con la base de datos de ventas globales..."):
    # La primera vez tardará 3 segundos. Las siguientes será instantáneo.
    df_ventas = cargar_datos_ventas()

st.write("Datos cargados. (Prueba a interactuar con la app, verás que la carga ya no toma 3 segundos).")
st.dataframe(df_ventas)
````

## 6.4 Control de Estado, Caché y Carga

### Estado Visual: Spinners y Progresos
Muestran empatía por el usuario cuando el proceso tardará más de 1.5 segundos.

### Rendimiento: `@st.cache_data` vs `@st.cache_resource`
- `@st.cache_data`: Serializa resultados de funciones matemáticas y bases de datos. Si una consulta SQL lenta retorna una tabla, Streamlit guardará la tabla final y no ejecutará el SQL las siguientes veces.
- `@st.cache_resource`: Se usa globalmente para Objetos Conexión (ej. MySQLConnection) o modelos de inferencias (pytorch, tensorflow). Guardarlos en resource evita que cada usuario cree una nueva sesión de BD en el pool.



````python
import streamlit as st
import time
import pandas as pd

st.header("Indicadores y Caché Avanzada")

# Función simulada costosa
@st.cache_data
def generar_reporte_complejo(anio):
    # Sin cache_data, si cambiáramos otro slider en la app, esta función
    # se re-ejecutaría haciéndonos esperar siempre 3 segundos.
    time.sleep(3) 
    return pd.DataFrame({"Mes": ["Ene", "Feb", "Mar"], "KPI": [100, 150, 130]})

if st.button("Ejecutar Consulta Financiera Lenta"):
    with st.spinner('Procesando millones de filas en el servidor...'):
        df_resultado = generar_reporte_complejo(2024)
        
    st.success("Reporte generado de caché o calculado nuevamente.")
    st.dataframe(df_resultado)
````

# Módulo 7: Aplicaciones Multipágina y Arquitectura Final

Ha llegado el momento de unir todas las piezas. Streamlit soporta la creación nativa de aplicaciones completas con múltiples páginas, simplemente estructurando correctamente la carpeta de nuestro proyecto.

## 7.1 Arquitectura Multipágina

Para que Streamlit genere automáticamente un menú lateral de navegación, debemos colocar nuestros scripts de página dentro de un directorio llamado `pages/`, junto a nuestro archivo principal `app.py`.

```text
miapp/
├── app.py                     (Página de inicio y declaración de estado global)
├── .streamlit/
│   └── config.toml            (Directivas de diseño primario)
├── pages/
│   ├── 01_Galeria_Interactiva.py (Subpágina 1)
│   ├── 02_Registro_Obras.py      (Subpágina 2)
│   └── 03_Analisis_Carga.py      (Subpágina 3)
├── utils/
│   ├── data_sim.py            (Lógica y datos compartidos)
│   └── estilos.css            (Archivo centralizado de hojas de estilo)
└── requirements.txt
```

### Ventajas de esta estructura
- **Mantenibilidad:** Separamos la interfaz en archivos lógicos distintos, haciéndolo fácil de entender y de trabajar en equipo.
- **Navegación nativa:** Streamlit lee prefijos numéricos en los nombres de archivo para ordenar el menú lateral automáticamente (ej. `01_` va antes de `02_`).

## 7.2 Preparación para Despliegue en Streamlit Community Cloud

A diferencia de Flask o Django, lanzar una app Streamlit en un VPS es complicado si no se manejan contenedores Docker o proxies inversos (Nginx/WebSockets).
La solución estándar y gratuita es usar GitHub + Streamlit Cloud.

**Reglas Críticas de Despliegue:**
1. Siempre incluye el archivo `requirements.txt`. Aquí enlistarás `streamlit`, `pandas`, `plotly` y cualquier otra libertad.
2. Si cometes el error de omitirlo, la aplicación lanzará un error *"ModuleNotFoundError"* tan pronto se lea el primer `import pandas`.
3. Para asegurar paths, jamás uses directorios absolutos como `C:/usuarios/docs/datos.csv`, usa directorios relativos en código como `./datos.csv` o `os.path.join()`.

## 7.3 Archivo de Entorno (`requirements.txt`)

Es fundamental generar este archivo listando todas las bibliotecas de las que depende nuestro proyecto para que, cuando realicemos el despliegue en *Streamlit Community Cloud* u otro servicio PaaS (Platform as a Service), el servidor sepa qué paquetes instalar.

`requirements.txt`
```text
streamlit>=1.34.0
```
