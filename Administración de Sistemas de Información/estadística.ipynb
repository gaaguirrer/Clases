# Introducción a NumPy
***
NumPy es una librería esencial para la computación numérica en Python. Proporciona soporte para arreglos multidimensionales (arrays) y una variedad de funciones matemáticas para operar con estos arreglos de manera eficiente. `pip intall numpy`

 **Funciones útiles de NumPy (con ejemplos)**
* `np.array()`: Crea un arreglo NumPy a partir de una lista o tupla.

````python
import numpy as np
arreglo = np.array([1, 2, 3, 4, 5])
print("Arreglo NumPy:", arreglo)
````

* `np.arange()`: Crea un arreglo NumPy con valores dentro de un rango determinado.

````python
arreglo_rango = np.arange(0, 10, 2)  # [0, 2, 4, 6, 8]
print("Arreglo con rango:", arreglo_rango)
````

* `np.linspace()`: Crea un arreglo NumPy con valores espaciados uniformemente dentro de un rango determinado.

````python
arreglo_espaciado = np.linspace(0, 1, 5)  # [0.  0.25 0.5  0.75 1.  ]
print("Arreglo espaciado:", arreglo_espaciado)
````

* `np.zeros()`: Crea un arreglo NumPy lleno de ceros.

````python
arreglo_ceros = np.zeros(5)  # [0. 0. 0. 0. 0.]
print("Arreglo de ceros:", arreglo_ceros)
````

* `np.ones()`: Crea un arreglo NumPy lleno de unos.

````python
arreglo_unos = np.ones(5)  # [1. 1. 1. 1. 1.]
print("Arreglo de unos:", arreglo_unos)
````

* `np.random.rand()`: Crea un arreglo NumPy con números aleatorios.

````python
arreglo_aleatorio = np.random.rand(5)
print("Arreglo aleatorio:", arreglo_aleatorio)
````

* `np.sum()`: Calcula la suma de los elementos de un arreglo.

````python
suma = np.sum(arreglo)
print("Suma:", suma)
````

* `np.mean()`: Calcula el promedio de los elementos de un arreglo.

````python
promedio = np.mean(arreglo)
print("Promedio:", promedio)
````

* `np.std()`: Calcula la desviación estándar de los elementos de un arreglo.

````python
desviacion_estandar = np.std(arreglo)
print("Desviación estándar:", desviacion_estandar)
````

* `np.median()`: Calcula la mediana de los elementos de un arreglo.

````python
mediana = np.median(arreglo)
print("Mediana:", mediana)
````

# Estadística Descriptiva con Pandas
Pandas proporciona funciones para calcular medidas estadísticas descriptivas como la media, la mediana, la moda, la varianza y la desviación estándar.

**Medidas de tendencia central (con ejemplos)**

* **Media:** Promedio de los valores.

````python
media_produccion = df_produccion['Producción (toneladas)'].mean()
print("Media de producción:", media_produccion)
````

* **Mediana:** Valor central de los valores ordenados.

````python
mediana_produccion = df_produccion['Producción (toneladas)'].median()
print("Mediana de producción:", mediana_produccion)
````

* **Moda:** Valor que aparece con mayor frecuencia.

````python
moda_produccion = df_produccion['Producción (toneladas)'].mode()
print("Moda de producción:\n", moda_produccion)
````

# **Medidas de dispersión**

 * **Varianza:** Promedio de los cuadrados de las diferencias entre los valores y la media.

````python
varianza_produccion = df_produccion['Producción (toneladas)'].var()
print("Varianza de producción:", varianza_produccion)
````

* **Desviación estándar:** Raíz cuadrada de la varianza.

````python
desviacion_estandar_produccion = df_produccion['Producción (toneladas)'].std()
print("Desviación estándar de producción:", desviacion_estandar_produccion)
````

 * **Rango**: diferencia entre el valor máximo y mínimo.

````python
rango_produccion = df_produccion['Producción (toneladas)'].max() - df_produccion['Producción (toneladas)'].min()
print("Rango de producción:", rango_produccion)
````

* **Cuartiles**: Valores que dividen los datos ordenados en cuatro partes iguales.

````python
cuartil_1 = df_produccion['Producción (toneladas)'].quantile(0.25)
cuartil_3 = df_produccion['Producción (toneladas)'].quantile(0.75)
print(f"Cuartil 1: {cuartil_1}, Cuartil 3: {cuartil_3}")
````

* **Percentiles**: Valores que dividen los datos ordenados en cien partes iguales.

````python
percentil_25 = df_produccion['Producción (toneladas)'].quantile(0.25)
percentil_75 = df_produccion['Producción (toneladas)'].quantile(0.75)
print(f"Percentil 25: {percentil_25}, Percentil 75: {percentil_75}")
````

## ¿Qué es la Regresión Lineal?

La regresión lineal es un método para encontrar la relación entre dos variables. Imagina que tienes una tabla con el tamaño de diferentes casas y sus precios. Queremos dibujar una línea recta que mejor represente esta relación, permitiéndonos predecir el precio de una casa dado su tamaño.

## Atributos y Etiquetas

* **Atributo (x1):** La característica que usamos para hacer la predicción (tamaño de la casa en m²).
* **Etiqueta (y'):** El valor que queremos predecir (precio de la casa en miles de dólares).

## La Fórmula Mágica: y' = b + w1x1

Esta fórmula nos permite calcular el precio predicho de una casa:

* **y':** Precio predicho.
* **b (Sesgo):** Punto donde la línea cruza el eje del precio.
* **w1 (Peso):** Pendiente de la línea, indica cuánto cambia el precio por cada m² adicional.
* **x1:** Tamaño de la casa.

## Ejemplo Práctico en Python

Vamos a generar datos de ejemplo y construir un modelo de regresión lineal. debemos instalar scikit learn `pip install --exists-action=i scikit-learn`, en caso de que se presenten problemas para reconocer la librería actualice `pip install --upgrade scikit-learn`

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Datos de ejemplo
tamaños = np.array([50, 75, 100, 125, 150, 175, 200, 225, 250, 275])
precios = np.array([150, 225, 300, 375, 450, 525, 600, 675, 750, 825])

# Graficamos los datos
plt.scatter(tamaños, precios)
plt.xlabel("Tamaño (metros cuadrados)")
plt.ylabel("Precio (miles de dólares)")
plt.title("Relación entre tamaño y precio")
plt.show()

# Creamos y entrenamos el modelo
modelo = LinearRegression()
modelo.fit(tamaños.reshape(-1, 1), precios)

# Obtenemos el sesgo y el peso
b = modelo.intercept_
w1 = modelo.coef_[0]

print("Sesgo (b):", b)
print("Peso (w1):", w1)

# Predicción para una casa de 160 metros cuadrados
tamaño_nueva_casa = 160
precio_predicho = modelo.predict([[tamaño_nueva_casa]])
print("Precio predicho para una casa de 160 m²:", precio_predicho[0])

# Graficamos la línea de regresión
puntos_regresion = b + w1 * tamaños
plt.scatter(tamaños, precios)
plt.plot(tamaños, puntos_regresion, color="red")
plt.xlabel("Tamaño (metros cuadrados)")
plt.ylabel("Precio (miles de dólares)")
plt.title("Regresión lineal: tamaño vs. precio")
plt.show()
````

### Explicación del Código

1.  **Datos:** Creamos arrays con tamaños y precios de casas.
2.  **Gráfico de dispersión:** Visualizamos la relación entre tamaño y precio.
3.  **Modelo:** Usamos `LinearRegression` para crear y entrenar el modelo.
4.  **Sesgo y peso:** Obtenemos los valores de `b` y `w1`.
5.  **Predicción:** Predecimos el precio de una casa nueva.
6.  **Línea de regresión:** Graficamos la línea que mejor se ajusta a los datos.

### Conceptos Clave

* **Entrenamiento:** Encontrar los valores óptimos de `b` y `w1`.
* **Predicción:** Usar la fórmula para estimar precios.
* **Visualización:** Graficar la línea de regresión.

## Determinación de la Ecuación Predictiva en Regresión Lineal Simple

Ahora aprenderemos cómo utilizar Python para determinar los valores de b (sesgo) y w1 (peso) en un modelo de regresión lineal simple. Estos valores nos permitirán crear una ecuación predictiva para estimar el precio de una casa basándonos en su tamaño.

### 1. Preparación de los Datos

Primero, necesitamos un conjunto de datos que relacione el tamaño de las casas con sus precios. Para este ejemplo, utilizaremos datos simulados.

````python
import numpy as np

# Tamaño de las casas (metros cuadrados)
tamaños = np.array([50, 75, 100, 125, 150, 175, 200, 225, 250, 275])

# Precios de las casas (miles de dólares)
precios = np.array([150, 225, 300, 375, 450, 525, 600, 675, 750, 825])
````

### 2. Entrenamiento del Modelo con scikit-learn

Utilizaremos la biblioteca scikit-learn para entrenar un modelo de regresión lineal. El modelo ajustará una línea recta a los datos y determinará los valores óptimos de b y w1.

````python
from sklearn.linear_model import LinearRegression

# Creamos el modelo
modelo = LinearRegression()

# Ajustamos el modelo a los datos
modelo.fit(tamaños.reshape(-1, 1), precios)
````

### 3. Obtención de los Valores de b y w1

Una vez que el modelo está entrenado, podemos obtener los valores de b (sesgo) y w1 (peso) utilizando los atributos `intercept_` y `coef_` del modelo.

### 4. Creación de la Ecuación Predictiva

Con los valores de b y w1, podemos crear la ecuación predictiva:

* y' = b + w1 * x1

Donde:

* y' es el precio predicho de la casa.
* x1 es el tamaño de la casa.

````python
# Obtenemos el sesgo (b)
b = modelo.intercept_

# Obtenemos el peso (w1)
w1 = modelo.coef_[0]

print("Sesgo (b):", b)
print("Peso (w1):", w1)
````

### 5. Uso de la Ecuación para Hacer Predicciones

Podemos utilizar la ecuación predictiva para estimar el precio de una casa dado su tamaño.

### 6. Función para la Ecuación Predictiva

Para mayor comodidad, podemos crear una función que encapsule la ecuación predictiva.

````python
# Tamaño de una casa nueva
tamaño_nueva_casa = 160

# Predicción del precio
precio_predicho = b + w1 * tamaño_nueva_casa

print("Precio predicho para una casa de", tamaño_nueva_casa, "metros cuadrados:", precio_predicho)

def predecir_precio_casa(tamaño):
    """
    Predice el precio de una casa dado su tamaño.

    Args:
        tamaño (float): Tamaño de la casa en metros cuadrados.

    Returns:
        float: Precio predicho de la casa en miles de dólares.
    """
    return b + w1 * tamaño

# Ejemplo de uso de la función
precio_predicho_funcion = predecir_precio_casa(160)
print("Precio predicho (usando función):", precio_predicho_funcion)
````

### Explicación Detallada

* **Sesgo (b):** Representa el valor predicho de la etiqueta (precio) cuando el atributo (tamaño) es cero.
* **Peso (w1):** Representa el cambio en la etiqueta (precio) por cada unidad de cambio en el atributo (tamaño).
